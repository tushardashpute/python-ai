This notebook demonstrates sampling behavior and randomness with NumPy. Sections:

2. Create Random Arrays (1D, 2D, 3D)
3. Seed the RNG
4. Generate Random Integers

```python
# 1) Import Required Libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# For simple unit tests
import unittest

print('numpy version:', np.__version__)
```

```python
# 2) Create Random Arrays (1D, 2D, 3D)
arr1 = np.random.rand(10)                # 1D float
arr2 = np.random.randint(0, 10, size=(4,5))  # 2D ints
arr3 = np.random.random(size=(2,3,4))   # 3D floats

print('arr1 shape', arr1.shape)
print('arr2 shape', arr2.shape)
print('arr3 shape', arr3.shape)
```

```python
# 3) Seed the RNG
np.random.seed(42)
arr_seed_a = np.random.rand(5)
np.random.seed(42)
arr_seed_b = np.random.rand(5)

print('Reproducible equal arrays? ', np.allclose(arr_seed_a, arr_seed_b))

# Generator API
rng = np.random.default_rng(42)
print('Generator sample:', rng.random(3))
```

```python
# 4) Generate Random Integers
ints = np.random.randint(low=0, high=100, size=(3,4))
print(ints)

# Generator API integers
rng = np.random.default_rng(123)
print('rng integers:', rng.integers(0, 10, size=5))
```

```python
# 5) Random Floats & Probabilities
floats = np.random.random(size=1000)
probs = floats / floats.sum()
print('floats mean:', floats.mean())
print('probs sum (≈1):', probs.sum())
```

```python
# 6) Sample from Common Distributions
normal = np.random.normal(loc=0, scale=1, size=1000)
binomial = np.random.binomial(n=10, p=0.5, size=1000)
uniform = np.random.random(size=1000)

print('normal mean ~0:', normal.mean())
print('binomial mean ~5:', binomial.mean())
```

```python
# 7) Save and Load Arrays
np.save('test_array.npy', normal)
loaded = np.load('test_array.npy')
print('loaded shape', loaded.shape)

np.savetxt('test_ints.csv', ints, delimiter=',')
print('wrote test_ints.csv')
```

```python
# 8) Basic Visualizations
plt.figure(figsize=(8,4))
sns.histplot(normal, kde=True)
plt.title('Normal distribution sample')
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(ints, annot=True, fmt='d')
plt.title('2D int matrix heatmap')
plt.show()
```

```python
# 9) Unit Tests for Reproducibility and Shape Checks
class TestRandomness(unittest.TestCase):
    def test_seed_repro(self):
        np.random.seed(123)
        a = np.random.rand(5)
        np.random.seed(123)
        b = np.random.rand(5)
        self.assertTrue(np.allclose(a, b))

    def test_shapes(self):
        self.assertEqual(ints.shape, (3,4))

# Run tests
unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
